```python
mmd = pd.read_parquet(f"{config.path_to_evaluation}/mmd_distances")
mmd  
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Set order of pot target sizes 
p_order = [5000, 12000, 25000, 50000, 100000, "all"]

mmd["p"] = mmd["p"].astype(str)

p_order_str = [str(x) for x in p_order]

# Create figure
fig, ax = plt.subplots(figsize=(10, 6))

# One line per combination of run and div
for (run, div), group in mmd.groupby(["run", "div"]):
    group = group.copy()

    group["p"] = pd.Categorical(
        group["p"],
        categories=p_order_str,
        ordered=True
    )

    group = group.sort_values("p")

    ax.plot(
        range(len(group)),
        group["mmd"],
        marker="o",
        label=f"{run}, div={div}"
    )

ax.set_xticks(range(len(p_order_str)))
ax.set_xticklabels(p_order_str)

ax.set_xlabel("p")
ax.set_ylabel("MMD")
ax.set_title("MMD accross similarity-level p")
ax.grid(True, alpha=0.3)

ax.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```
